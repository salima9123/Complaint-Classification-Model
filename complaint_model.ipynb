```python
print("Available columns:")
for column in df.columns:
    print("-", repr(column))

# Look for likely complaint-text columns
text_candidates = [
    column for column in df.columns
    if any(word in column.lower() for word in [
        "complaint", "description", "issue", "problem",
        "feedback", "details", "text", "message"
    ])
]

# Look for likely category-label columns
category_candidates = [
    column for column in df.columns
    if any(word in column.lower() for word in [
        "category", "type", "classification", "department",
        "issue_type", "complaint_type"
    ])
]

print("\nPossible complaint columns:", text_candidates)
print("Possible category columns:", category_candidates)

# Explicitly setting the correct columns to resolve the conflict
complaint_column = "issue_description"
category_column = "issue_type"

print("\nSelected complaint column:", complaint_column)
print("Selected category column:", category_column)

if complaint_column is None or category_column is None:
    raise ValueError(
        "Set complaint_column and category_column to the exact "
        "column names shown above, then rerun this cell."
    )

if complaint_column == category_column:
    raise ValueError("Complaint and category columns must be different.")

print("\nComplaint examples:")
display(df[[complaint_column, category_column]].head(10))
```

```python
print("Rows before cleaning:", len(df))

# Keep only the columns required for classification
data = df[[complaint_column, category_column]].copy()

# Remove missing complaint text and missing category labels
data = data.dropna(subset=[complaint_column, category_column])

# Convert complaint text to strings and remove extra spaces
data[complaint_column] = (
    data[complaint_column].astype(str).str.strip()
)

# Remove empty complaint text
data = data[data[complaint_column] != ""]

# Remove rows containing only whitespace
data = data[
    data[complaint_column].str.strip().ne("")
]

# Clean category labels
data[category_column] = (
    data[category_column].astype(str).str.strip()
)

data = data[data[category_column] != ""]

# Remove duplicate complaint-category pairs
data = data.drop_duplicates(
    subset=[complaint_column, category_column]
).reset_index(drop=True)

print("Rows after cleaning:", len(data))
print("Rows removed:", len(df) - len(data))

print("\nCategory distribution:")
print(data[category_column].value_counts())

print("\nCleaned dataset:")
display(data.head())
```

```python
from sklearn.model_selection import train_test_split

Features = data[complaint_column]
Labels = data[category_column]

print("Features shape:", Features.shape)
print("Labels shape:", Labels.shape)

# Use stratification when every category has at least 2 records
stratify_labels = (
    Labels if Labels.value_counts().min() >= 2 else None
)

if stratify_labels is None:
    print(
        "Warning: At least one category has fewer than 2 records. "
        "Using a non-stratified split."
    )

Features_train, Features_test, Labels_train, Labels_test = (
    train_test_split(
        Features,
        Labels,
        test_size=0.20,
        random_state=42,
        stratify=stratify_labels
    )
)

print("\nTraining records:", len(Features_train))
print("Testing records:", len(Features_test))
```

```python
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.linear_model import LogisticRegression

# Build the complete machine learning pipeline
complaint_model = Pipeline([
    ("vectorizer", CountVectorizer()),
    ("classifier", LogisticRegression(max_iter=1000))
])

# Train the model
complaint_model.fit(Features_train, Labels_train)

print("Logistic Regression model trained successfully!")
print("Pipeline:", complaint_model)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    ConfusionMatrixDisplay
)
import matplotlib.pyplot as plt

# Predict categories for the test dataset
Labels_pred = complaint_model.predict(Features_test)

# Calculate evaluation metrics
accuracy = accuracy_score(Labels_test, Labels_pred)

precision = precision_score(
    Labels_test, Labels_pred,
    average="weighted",
    zero_division=0
)

recall = recall_score(
    Labels_test, Labels_pred,
    average="weighted",
    zero_division=0
)

f1 = f1_score(
    Labels_test, Labels_pred,
    average="weighted",
    zero_division=0
)

print("Model Evaluation Results")
print("------------------------")
print(f"Accuracy:  {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall:    {recall:.4f}")
print(f"F1-score:  {f1:.4f}")

print("\nClassification Report:")
print(classification_report(
    Labels_test,
    Labels_pred,
    zero_division=0
))

print("Confusion Matrix:")

ConfusionMatrixDisplay.from_predictions(
    Labels_test,
    Labels_pred,
    xticks_rotation="vertical",
    cmap="Blues"
)

plt.title("Complaint Classification Confusion Matrix")
plt.tight_layout()
plt.show()
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output

# Text box for entering one or multiple complaints
complaint_input = widgets.Textarea(
    value="",
    placeholder=(
        "Enter a complaint here...\n"
        "Example: Garbage is not collected in my street.\n"
        "Example: The streetlight is not working."
    ),
    description="Complaint:",
    layout=widgets.Layout(width="100%", height="150px"),
    style={"description_width": "initial"}
)

# Prediction button
predict_button = widgets.Button(
    description="Predict Categories",
    button_style="primary",
    icon="check"
)

# Output area
prediction_output = widgets.Output()

def predict_complaints(button):
    with prediction_output:
        clear_output(wait=True)

        # Split input into separate complaints, one per line
        complaints = [
            line.strip()
            for line in complaint_input.value.splitlines()
            if line.strip()
        ]

        if not complaints:
            print("Please enter at least one complaint.")
            return

        # Use the previously loaded model; do not retrain
        predictions = loaded_model.predict(complaints)

        print("Complaint Classification Results\n")

        for index, (complaint, category) in enumerate(
            zip(complaints, predictions), start=1
        ):
            print(f"Complaint {index}:")
            print("Entered complaint:", complaint)
            print("Predicted category:", category)
            print("-" * 50)

predict_button.on_click(predict_complaints)

display(complaint_input)
display(predict_button)
display(prediction_output)
```
